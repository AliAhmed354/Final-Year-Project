```python
df = dataf.drop(['Patient ID ','Moth of Birth', 'Sex', 'Marital Status', 'Occupation Code', 'Adopted Status'], axis=1, inplace = False )
df.head()
```

```python
df = df.dropna()
df.isnull().sum()
```

```python
df['T Stage '].unique()
df['N Stage'].unique()
df['differentiate'].unique()
df['A Stage'].unique()
```

```python
df['T Stage ']= df['T Stage '].map({'T1':1, 'T2':2, 'T3':3, 'T4':4})
```

```python
df['N Stage'] = df['N Stage'].map({'N1':1, 'N2':2, 'N3':3})
```

```python
df['differentiate'] = df['differentiate'].map({'Poorly differentiated':1, 'Moderately differentiated':2,'Well differentiated':3, 'Undifferentiated':4})

```

```python
df['A Stage'] = df['A Stage'].map({'Regional':1, 'Distant':2})
```

```python
df['Estrogen Status'] = df['Estrogen Status'].map({'Negative':1, 'Positive':2})
```

```python
df['Progesterone Status'] = df['Progesterone Status'].map({'Negative':1, 'Positive':2})
```

```python
df['6th Stage'] = df['6th Stage'].map({'IIA':1, 'IIB':2, 'IIIA':3, 'IIIB':4, 'IIIC':5 })
```

```python
df['Mortality_Status'] = df['Mortality_Status'].map({'Alive':0, 'Dead':1})
```

```python
from sklearn import preprocessing
label_encoder = preprocessing.LabelEncoder()

df['Race'] = label_encoder.fit_transform(df['Race'])
```

```python
df[['Regional Node Examined', 'Reginol Node Positive']] = df[['Regional Node Examined', 'Reginol Node Positive']].apply(pd.to_numeric,errors='coerce')
df.dropna(inplace=True)
df[['Regional Node Examined', 'Reginol Node Positive']] = df[['Regional Node Examined', 'Reginol Node Positive']].apply(pd.to_numeric,errors='coerce')

df.dropna(inplace=True)
df.info()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

cols = ['Age', 'Tumor Size', 'Regional Node Examined', 'Reginol Node Positive', 'Survival Months']

q1 = df[cols].quantile(0.25)
q3 = df[cols].quantile(0.75)
iqr = q3 - q1

lower = q1 - 5 * iqr
upper = q3 + 5 * iqr

outliers = ~((df[cols] < lower) | (df[cols] > upper)).any(axis=1)

newdf = df[outliers]

newdf = newdf[newdf['Tumor Size'] !=-75]

newdf.describe()

```

```python
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score, confusion_matrix
## defining x and y vales, train test split

xvalues = newdf.drop(columns=['Mortality_Status'])
yvalues = newdf['Mortality_Status']

xtrain, xtest, ytrain, ytest = train_test_split(
    xvalues, yvalues, test_size=0.2, random_state=42, stratify=yvalues
)

## set models with parameters

models = {
    'rforrest': Pipeline([
        ('scaler', StandardScaler()),
        ('model', RandomForestClassifier(n_estimators=100, random_state=42))
    ]),

    'svm': Pipeline([
        ('scaler', StandardScaler()),
        ('model', SVC(kernel='rbf', random_state=42))
    ]),

    'knn': Pipeline([
        ('scaler', StandardScaler()),
        ('model', KNeighborsClassifier(n_neighbors=10))
    ])
}

## set resampling techniques

from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from imblearn.combine import SMOTETomek

resamplers = {
    'none': None,
    'smote':SMOTE(random_state=42),
    'unders':RandomUnderSampler(random_state=42),
    'smotetomek':SMOTETomek(random_state=42)
}


```

```python
## looping through each combo

results = []
confmatrix = []

count = 0

for rname, resampler in resamplers.items():
  for mname, pipeline in models.items():

    xt_result, yt_result = xtrain, ytrain

    if resampler is not None:
      xt_result, yt_result = resampler.fit_resample(xtrain, ytrain)

    pipeline.fit(xt_result, yt_result)
    ypred = pipeline.predict(xtest)

    f1 = f1_score(ytest, ypred)
    acc = accuracy_score(ytest, ypred)

    tn, fp, fn, tp = confusion_matrix(ytest, ypred).ravel()

    results.append({
        'count': count,
        'resampler':rname,
        'model':mname,
        'f1':f1,
        'accuracy':acc})

    confmatrix.append({
        'count' : count,
        'resampler':rname,
        'model':mname,
        'tn':tn,
        'fp':fp,
        'fn':fn,
        'tp':tp
    })

    count += 1





```

```python
resultsdf = pd.DataFrame(results)
resultsdf.head(20)
```
